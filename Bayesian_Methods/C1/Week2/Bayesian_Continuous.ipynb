```python
#Bayesian inference - Continuous Prior, Discrete Likelihood

#Define the prior distribution pi(sigma | data) as a beta distribution

from scipy.stats import beta

a, b = 5, 2
rv = beta(a,b)

prior = lambda x: rv.pdf(x)
```

```python
#Now define the likelihood
#The Binomial likelihood is interpreted as Bernoulli ~ p^16*(1-p)^(20-16)

from math import factorial

x1 = int(20)
x2 = int(16)

c = factorial(x1)/(factorial(x2)*factorial(x1-x2))

l = lambda x: c*(x**x2)*((1-x)**(x1-x2))
```

```python
#Now we compute the marginal likelihood by integrating the product of the likelihood and prior and integrating out the parameter sigma
#sigma is integrated from 0 to 1, the full range of the success rate for one trial
#We compute f(data)

import scipy.integrate as integrate

f = lambda x: prior(x)*l(x)

tp = integrate.quad(f, 0, 1)[0]
```

```python
#Now compute the success rate as being greater than or equal to 80%, ie. compute Pr[sigma >= 80%]

#Define the posterior distribution : pi(sigma | data)

post = lambda x : prior(x)*l(x)/tp

#Get the probability by integrating the posterior distribution for sigma between 0.8 and 1.0

print(integrate.quad(post, float(0.8), float(1.0))[0])
```
