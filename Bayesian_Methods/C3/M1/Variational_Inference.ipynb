```python
import numpy as np
import pandas as pd
import pymc as pm
import arviz as az

# ---------------------------------------------------------
# 1. Example churn dataset
# ---------------------------------------------------------

np.random.seed(42)
n = 10000

df = pd.DataFrame({
    "mpr": np.random.beta(5, 2, n),
    "oop": np.random.gamma(shape=2.5, scale=35, size=n),
    "qm": np.random.binomial(1, 0.25, n),
})

df["high_oop"] = (df["oop"] > 113).astype(int)

# Standardize OOP for stable inference
df["oop_std"] = (
    df["oop"] - df["oop"].mean()
) / df["oop"].std()

# ---------------------------------------------------------
# 2. Generate illustrative churn outcome
# ---------------------------------------------------------

logit_p = (
    -0.5
    - 2.0 * df["mpr"]
    + 0.20 * df["oop_std"]
    + 0.35 * df["qm"]
    + 0.45 * df["high_oop"]
    + 0.50 * df["qm"] * df["high_oop"]
)

p = 1 / (1 + np.exp(-logit_p))

df["churn"] = np.random.binomial(1, p)

# ---------------------------------------------------------
# 3. Bayesian logistic regression
# ---------------------------------------------------------

with pm.Model() as churn_model:

    beta0 = pm.Normal("beta0", 0, 2)

    beta_mpr = pm.Normal("beta_mpr", 0, 1)
    beta_oop = pm.Normal("beta_oop", 0, 1)
    beta_qm = pm.Normal("beta_qm", 0, 1)
    beta_high_oop = pm.Normal("beta_high_oop", 0, 1)
    beta_qm_high_oop = pm.Normal("beta_qm_high_oop", 0, 1)

    eta = (
        beta0
        + beta_mpr * df["mpr"].values
        + beta_oop * df["oop_std"].values
        + beta_qm * df["qm"].values
        + beta_high_oop * df["high_oop"].values
        + beta_qm_high_oop
        * df["qm"].values
        * df["high_oop"].values
    )

    p_churn = pm.math.sigmoid(eta)

    y_obs = pm.Bernoulli(
        "y_obs",
        p=p_churn,
        observed=df["churn"].values
    )

    # -----------------------------------------------------
    # 4. Variational inference
    # -----------------------------------------------------

    approx = pm.fit(
        n=20_000,
        method="advi",
        random_seed=42
    )

# ---------------------------------------------------------
# 5. Draw from variational posterior
# ---------------------------------------------------------

posterior = approx.sample(
    draws=5000,
    random_seed=42
)

print(az.summary(posterior))
```
